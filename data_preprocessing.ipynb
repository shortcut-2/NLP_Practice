### Word Embedding
- 참고용 메모: 불용어 더 넣어야 함 데이터가 덜 빠져서 학습이 덜 됨
    - 면접이라 답변이 긍정적 단어만 있을 것 같아서 고민 중
    - QnA를 합쳐서 관리해야 하는지도 의문

```python
import re

# 데이터 로드
df = pd.read_csv('./data/raw_data/merged_data.csv')

# 결측치 확인
df.isnull().sum()
```

```python
# 한글 아닌 데이터 제거
df['question_raw'] = df['question_raw'].replace(r'[^0-9가-힣ㄱ-ㅎㅏ-ㅣ\s]', '', regex=True)
df['answer_raw'] = df['answer_raw'].replace(r'[^0-9가-힣ㄱ-ㅎㅏ-ㅣ\s]', '', regex=True)
```

```python
from tqdm import tqdm
# def load_ko_stopwords(filepath):
#     with open(filepath, 'r', encoding='UTF-8') as f:
#         return [line.strip() for line in f]

def preprocess(sentences, ko_stopwords):
    result = []
    for sentence in tqdm(sentences):
        tokens = okt.morphs(sentence, stem=True)
        tokens = [token for token in tokens if token not in ko_stopwords]
        result.append(tokens)
    return result

ko_stopwords = load_ko_stopwords('./ko_stopwords.txt')
preprocessed_q_data = preprocess(df['question_raw'], ko_stopwords)
preprocessed_a_data = preprocess(df['answer_raw'], ko_stopwords)
```

```python
# 모델 학습 (qna 합쳐서 할지 분리해서 할지 고민)
from gensim.models import Word2Vec

model = Word2Vec(
    sentences=preprocessed_q_data,
    vector_size=100,
    window=5,
    min_count=5,
    sg=1
)
```
