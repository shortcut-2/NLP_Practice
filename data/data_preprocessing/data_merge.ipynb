# Data merge

```python
import json
import glob
import pandas as pd
```

```python
file_path = '../raw_data/*.json' 
json_files = glob.glob(file_path)

data_list = []

for file in json_files:
    with open(file, 'r', encoding='utf-8') as f:
        try:
            content = f.read()
            data = json.loads(content, strict=False)
        
            # 데이터 추출
            info = data.get('dataSet', {}).get('info', {})
            question = data.get('dataSet', {}).get('question', {}).get('raw', {})
            answer = data.get('dataSet', {}).get('answer', {}).get('raw', {})
            
            # 딕셔너리로 추출
            extracted_data = {
                'date': info.get('date'),
                'occupation': info.get('occupation'),
                'channel': info.get('channel'),
                'place': info.get('place'),
                'gender': info.get('gender'),
                'ageRange': info.get('ageRange'),
                'question_raw': question.get('text'),
                'answer_raw': answer.get('text')
            }
        
            data_list.append(extracted_data)

        except json.JSONDecodeError as e:
            print(f"파일 읽기 실패: {file}")
            print(f"에러 메시지: {e}")
            continue
        
df = pd.DataFrame(data_list)

print(f"총 {len(df)}개의 파일을 합쳤습니다.")
```

```python
df = df.sort_values(['date', 'occupation','ageRange'], ignore_index=True)
```
